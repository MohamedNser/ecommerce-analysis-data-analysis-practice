```python
df = df.dropna(subset=['product_id', 'price', 'seller_id'])
```

```python
df['order_delivered_customer_date'] = df['order_delivered_customer_date'].fillna('Not Delivered')
df['order_delivered_carrier_date'] = df['order_delivered_carrier_date'].fillna('Not Delivered')
```

```python
df = df.dropna(subset=['payment_type', 'payment_value'])
```

```python
cols = ['product_category_name', 'product_name_lenght', 
        'product_description_lenght', 'product_photos_qty']
df[cols] = df[cols].fillna('Unknown')
```

```python
cols = ['product_weight_g', 'product_length_cm', 
        'product_height_cm', 'product_width_cm']
df[cols] = df[cols].fillna(df[cols].median())
```

```python
df = df.dropna(subset=['order_approved_at'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Top 10 Categories by Orders
top_categories = df.groupby('product_category_name')['order_id'].count().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_categories.values, y=top_categories.index, palette='Blues_r')
plt.title('Top 10 Product Categories by Orders')
plt.xlabel('Number of Orders')
plt.tight_layout()
plt.show()
```

```python
# 2. Monthly Revenue Over Time
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df['month'] = df['order_purchase_timestamp'].dt.to_period('M')

monthly_revenue = df.groupby('month')['price'].sum()

plt.figure(figsize=(14,6))
monthly_revenue.plot(kind='line', marker='o', color='blue')
plt.title('Monthly Revenue Over Time')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
# 3. Review Score Distribution
plt.figure(figsize=(8,5))
sns.countplot(x='review_score', data=df, palette='RdYlGn')
plt.title('Review Score Distribution')
plt.xlabel('Score')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# 4. Top 10 Cities
top_cities = df.groupby('customer_city')['order_id'].count().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_cities.values, y=top_cities.index, palette='Greens_r')
plt.title('Top 10 Cities by Orders')
plt.xlabel('Number of Orders')
plt.tight_layout()
plt.show()
```

```python
# 5. Payment Methods
payment_counts = df['payment_type'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(payment_counts.values, labels=payment_counts.index, autopct='%1.1f%%', 
        colors=sns.color_palette('Set2'))
plt.title('Payment Methods Distribution')
plt.tight_layout()
plt.show()
```
